```python
# Practice
from pandas import Series, DataFrame
import pandas as pd

attri_data = {'ID':['1', '2', '3', '4', '5'],
              'Sex':['F', 'F', 'M', 'M', 'F'],
              'Money':[1000, 2000, 500, 300, 700],
              'Name':['Saito', 'Horie', 'Kondo', 'Kawada', 'Matsubara']}
attri_data_frame1 = DataFrame(attri_data)
```

```python
# 練習問題１ 上のデータに対して、Moneyが500以上の人を絞り込んで、、レコードを表示
attri_data_frame1[attri_data_frame1['Money'] >= 500]
```

```python
# 練習問題２ データに対して、男女別の平均Moneyを求める
attri_data_frame1.groupby('Sex')['Money'].mean()
```

```python
# 練習問題３ 次のデータの同じIDの人をキーとして、データをマージする。そして、Money, Math と Englishの平均を求める
attri_data2 = {'ID':['3', '4', '7'],
               'Math':[60, 30, 40],
               'English':[80, 20, 30]}
attri_data_frame2 = DataFrame(attri_data2)

merge_data = attri_data_frame1.merge(attri_data_frame2)

print('Moneyの平均:',merge_data['Money'].mean())
print('Mathの平均:',merge_data['Math'].mean())
print('Englishの平均:', merge_data['English'].mean())
```

```python
# 1000個の一葉乱数を２組発生させるプログラム
import math

def uni_hist(N):
    # 一様乱数の発生
    x = np.random.uniform(0.0, 1.0, N)
    y = np.random.uniform(0.0, 1.0, N)

    plt.subplot(2, 1, 1)
    plt.hist(x)
    plt.title('No1:histogram')

    plt.grid(True)

    plt.subplot(2, 1, 2)
    plt.hist(y)
    plt.title('No2:hitsogram')

    plt.grid(True)

    # タイトルの被りを防ぐ
    plt.tight_layout()

uni_hist(100)
```
